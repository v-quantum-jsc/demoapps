## V-QUANTUM-ANNEALING được dùng để tính toán tối ưu hóa:

Ví dụ một bài toán tối ưu hóa sử dụng mô hình tối ưu dạng Ising như sau:

$$H(\{\sigma_i\}) = \sum_{i > j} J_{ij}\sigma_i \sigma_j + \sum_{i=1}^N h_i \sigma_i$$
$$\sigma_i \in \{-1, 1\}, i=1,\cdots N$$

Trong đó, $H(\{\sigma_i\})$ là hàm Hamitonion. Nó là hàm số biểu thị năng lượng hay hàm mục tiêu, hay hàm chi phí.
$\sigma_i$は$\{1, -1\}$の2値を取る変数です。

> $\sigma_i$ là biến số.
> Spin là các nam châm từ đơn lẻ. Khi nó bằng -1 thì nam châm hướng lên trên, khi nó bằng 1 thì nam châm hướng xuống dưới.


$H$là hàm số liên quan trực tiếp đến tập các biến số: $\{\sigma_i\} = \{\sigma_1, \sigma_2, \cdots, \sigma_N\}$.
$J_{ij}, h_i$ là các hệ số quan trọng, thể hiện ảnh hưởng tương hỗ giữa các biến số khác nhau, nó cũng hay được gọi là các từ trường theo phương nằm ngang.

V-QUANTUM-ANNEALING được tính toán như sau: khi các thông tin đầu vào (inputs) là các hệ số : $J_{ij}$と$h_i$ thì V-QUANTUM-ANNEALING sẽ được sử dụng như một thư viện tính toán, sử dụng lý thuyết lượng tử biểu thị bằng các phương pháp như : Simulated Annealing, Simulated Quantum Annealing để thực hiện nhiệm vụ tính toán làm sao cho tìm được giá trị nhỏ nhất của $H(\{\sigma_i\})$ với các cặp biến số tối ưu: $\{\sigma_i\}$.

$$ \begin{pmatrix}
h_{0} & J_{0,1} & \cdots & J_{0,N-1}\\

## Ví dụ sử dụng V-QUANTUM-ANNEALING để giải bài toán tối ưu bằng phương pháp QUBO như sau:
Trong các bài toán thực tế ta sẽ sử dụng phổ biến hơn bằng phương pháp QUBO(Quadratic unconstraited binary optimization).

QUBO được viết dưới dạng toán học như sau:

$$H(\{q_i\}) = \sum_{i\geq j} Q_{ij}q_i q_j$$
$$q_i \in \{0, 1\}$$

Điểm khác biệt cơ bản so với phương pháp ising là phương pháp QUBO chỉ sử dụng 2 biến số duy nhất là 0 và 1 để thể hiện và tính toán.$\sum Q_{ij}$ là tổng số các biến trong phạm vi tính toán( Ví dụ $Q_{ij}$ là một ma trận đối xứng) sẽ đượ viết dưới công thức toán học như sau:

> ta hoàn toàn có thể sử dụng công thức: $q_i = (\sigma_i + 1)/2$ để biến đổi hàm số từ phương pháp ising sang phương pháp QUBO, với ý nghĩa toán học như vậy thì có thể nói 2 phương pháp này là hoàn toàn tương đồng với nhau.

QUBO sẽ sử dụng các hệ số đầu vào là tập hợp: $Q_{ij}$ để nhập và tính toán tối ưu cho hàm số: $H(\{q_i\})$ với tập hợp các hệ số có giá trị tương ứng 0, 1, và việc phải làm là tìm các hệ số $\{q_i\}$ sao cho hàm số có giá trị nhỏ nhất. điều này hoàn toàn tương tự với phương pháp Ising.

Tiếp thep $q_i$ là hệ số dạng binary, $q_i^2 = q_i$ sẽ luôn đúng, do đó ta có thể viết hàm mục tiêu dưới dạng như sau:

$$H(\{q_i\}) = \sum_{i > j} Q_{ij}q_i q_j + \sum_i Q_{ii} q_i$$

$Q_{ij}$ có các chỉ số i,j tương ứng với $q_i$.

Nào, hay thử sử dụng V-QUANTUM-ANNEALING và QUBO để giải thử xem sao!

````python
# Tạo một bộ từ điển các hệ số Q_ij
Q = {(0, 0): -1, (0, 1): -1, (1, 2): 1, (2, 2): 1}
sampler = vqa.SASampler()
# để giải hàm QUBO ta sẽ sử dụng hàm số.sample_qubo như sau:
response = sampler.sample_qubo(Q)
print(response.states)
````

QUBO chỉ có giá trị 0, 1 nên đương nhiên là giá trị tối ưu cũng chỉ có dạng 0, 1. Điều đó có nghĩa là dù ta sử dụng Phương pháp Ising hay Phương pháp QUBO thì sẽ cho ra cùng một kế t quả tối ưu.

````python
# 実は h, J の添字を示す、辞書のkeyは数値以外も扱うことができます。
h = {'a': -1, 'b': -1}
J = {('a', 'b'): -1, ('b', 'c'): 1}
# num_reads 引数に値を代入することで、SQAを10回試行する計算を一度の命令で解くことができます。
sampler = vqa.SQASampler(num_reads=10)
response = sampler.sample_ising(h, J)
print(response.first.sample)
print(response.first.energy)
````

````python
# response.states を見てみましょう。10回分の解が入っていることがわかります。
print(response.states)
````

> num_reads などコンストラクタで渡すパラメータは`.sample_ising`などのサンプリングを実行するメソッドで上書きすることができます。
> ```
> response = sampler.sample_ising(h, J, num_reads=2)
> response.states
> > [[1, 1, -1],[1, 1, -1]]
> ```

今回は問題が簡単なので、10回とも同じ答え [1,1,-1] になっていることがわかります。

````python
# 次にエネルギーを見てみましょう。
response.energies
````

エネルギーの値を10回とも同じ値を取っていることがわかります。
`response.states`に入っている解はリストになっているため、問題をセットした時の `a, b, c`という文字列との対応がわかりません。それを調べるために`response.variables`を見てみましょう。

````python
response.variables
````

最小のエネルギー値を持った状態のみが知りたい場合には `.first` が便利です。

````python
response.first
````

## Tạo Random QUBO để tính toán tối ưu:

Hãy thử ví dụ với bài toán có 50 biến số, tạo các Randome cho tập hợp $Q_{ij}$ và chuyển sang QUBO sau đó giải:

````python
N = 50
# Tạo mới random một tập các hệ số Qij như sau:
import random
Q = {(i, j): random.uniform(-1, 1) for i in range(N) for j in range(i+1, N)}

# V-QUANTUM-ANNEALING để giải:
sampler = vqa.SASampler()
response = sampler.sample_qubo(Q, num_reads=100)
````

````python
# Tìm năng lượng nhỏ nhất như sau:
response.energies[:5]
````

Hãy xem chi tiết kết quả năng lượng được liệt kê ở trên ta sẽ thấy các kết quả khác nhau do ta đã sử dụng các hệ số ngẫu nhiên.
Do ta sử dụng Random để tạo $Q_{ij}$ cho bài toán, nên về mặt lý thuyết mỗi lần sẽ tạo ra một kết quả khác nhau.
Dưới đây sẽ vẽ kết quả đó dưới hạng Histogram như sau:

````python
import matplotlib.pyplot as plt
plt.hist(response.energies, bins=15)
plt.xlabel('Energy', fontsize=15)
plt.ylabel('Frequency', fontsize=15)
plt.show()
````

Năng lượng càng nhỏ thì trạng thái càng tốt, Trong hình histogram này thì ta sẽ vẽ cả các trạng thái lớn và nhỏ. Trong hình có thể thấy kết quả phần lớn là thu được trạng thái ở mức năng lượng nhỏ. Ta sẽ tìm xem là vị trí nào là tạo ra năng lượng nhỏ nhất, đó chính là nghiệm cận với nghiệm tối ưu toàn phần của bài toán, bằng cách sử dụng lệnh states như sau:
